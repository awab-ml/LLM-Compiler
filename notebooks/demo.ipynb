# LLMCompiler Interactive Demo

This notebook demonstrates the LLMCompiler architecture with interactive examples.

## Setup

```python
import sys
import os
from dotenv import load_dotenv

# Add parent directory to path
sys.path.insert(0, os.path.abspath('..'))

# Load environment variables
load_dotenv()

print("✓ Environment loaded")
```

## Import LLMCompiler

```python
from src.core.graph import create_llm_compiler_graph
from src.tools import get_all_tools

# Create the graph
graph = create_llm_compiler_graph()

print("✓ LLMCompiler graph created")
```

## View Available Tools

```python
tools = get_all_tools()

print("Available Tools:")
print("=" * 60)
for tool in tools:
    print(f"\n{tool.name}")
    print(f"  Description: {tool.description}")
```

## Example 1: Simple Search Query

```python
query = "What is the capital of France?"

result = graph.invoke({
    "messages": [{"role": "user", "content": query}],
    "tasks": [],
    "task_results": {},
    "should_replan": False,
    "replan_count": 0
})

print("Query:", query)
print("\nResponse:")
print("=" * 60)
print(result["messages"][-1].content)
```

## Example 2: Math Calculation

```python
query = "What is the square root of 144 multiplied by 5?"

result = graph.invoke({
    "messages": [{"role": "user", "content": query}],
    "tasks": [],
    "task_results": {},
    "should_replan": False,
    "replan_count": 0
})

print("Query:", query)
print("\nResponse:")
print("=" * 60)
print(result["messages"][-1].content)
```

## Example 3: Multi-Hop Reasoning

```python
query = "What is the combined population of Tokyo and New York City?"

result = graph.invoke({
    "messages": [{"role": "user", "content": query}],
    "tasks": [],
    "task_results": {},
    "should_replan": False,
    "replan_count": 0
})

print("Query:", query)
print("\nResponse:")
print("=" * 60)
print(result["messages"][-1].content)

print("\n\nExecution Details:")
print("=" * 60)
print(f"Tasks executed: {len(result.get('tasks', []))}")
print(f"Replanning iterations: {result.get('replan_count', 0)}")
```

## Example 4: Complex Query with Search and Math

```python
query = "If the GDP of California is $3.9 trillion, what would it be after a 20% increase?"

result = graph.invoke({
    "messages": [{"role": "user", "content": query}],
    "tasks": [],
    "task_results": {},
    "should_replan": False,
    "replan_count": 0
})

print("Query:", query)
print("\nResponse:")
print("=" * 60)
print(result["messages"][-1].content)
```

## Inspect Task Execution

```python
# View the tasks that were executed
if result.get('tasks'):
    print("Tasks Executed:")
    print("=" * 60)
    for task in result['tasks']:
        print(f"\nTask {task['idx']}: {task['tool']}")
        print(f"  Args: {task['args']}")
        print(f"  Dependencies: {task['dependencies']}")
    
    print("\n\nTask Results:")
    print("=" * 60)
    for idx, res in result.get('task_results', {}).items():
        print(f"\nTask {idx}:")
        print(f"  {res[:200]}...")  # Show first 200 chars
else:
    print("No tasks were executed (or tasks were cleared after completion)")
```
